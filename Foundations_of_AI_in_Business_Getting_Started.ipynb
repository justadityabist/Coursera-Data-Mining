## Step 1 — Create a Business Dataset

### Business scenario

Imagine that you are working as a business analyst for an **e-commerce company**.

You have been asked to create a synthetic dataset containing **500 customer orders**.

The dataset should contain at least the following variables:

- Order ID
- Customer age
- Customer location

- Discount percentage

- Whether the product was returned

### Your task

Open Gemini using the Gemini icon in Colab and ask it to generate the required Python code.

You may start with this prompt:

> **Create a synthetic dataset for an e-commerce company containing 500 customer orders. Include order ID, customer age, customer location, product category, order value, discount percentage, delivery time, customer rating, and whether the product was returned. Use realistic-looking synthetic data. Store the result as a pandas DataFrame called `df` and display the first 10 rows.**

Ask Gemini to generate the code and run the resulting code in the cell below.

### Code generated by Gemini

Paste the code generated by Gemini into the cell below and run it.


```python
import pandas as pd
import numpy as np

# Set random seed for reproducibility
np.random.seed(42)

# Generate 500 orders
n_orders = 500

# Synthetic data generation
order_ids = [f"ORD{10000 + i}" for i in range(n_orders)]
customer_ages = np.random.randint(18, 70, size=n_orders)
locations = np.random.choice(['New York', 'Los Angeles', 'Chicago', 'Houston', 'Phoenix', 'Philadelphia', 'San Antonio', 'San Diego'], size=n_orders)
categories = np.random.choice(['Electronics', 'Clothing', 'Home & Kitchen', 'Beauty', 'Books', 'Sports'], size=n_orders, p=[0.25, 0.30, 0.15, 0.15, 0.10, 0.05])
order_values = np.round(np.random.exponential(scale=80, size=n_orders) + 10, 2)
discounts = np.random.choice([0.0, 0.05, 0.10, 0.15, 0.20, 0.30, 0.50], size=n_orders, p=[0.40, 0.15, 0.15, 0.10, 0.10, 0.07, 0.03])
delivery_times = np.random.poisson(lam=4, size=n_orders) + 1 # At least 1 day
customer_ratings = np.random.choice([1, 2, 3, 4, 5], size=n_orders, p=[0.05, 0.08, 0.12, 0.35, 0.40])
returned = np.random.choice([True, False], size=n_orders, p=[0.15, 0.85])

# Create DataFrame
df = pd.DataFrame({
    'Order ID': order_ids,
    'Customer Age': customer_ages,
    'Customer Location': locations,
    'Product Category': categories,
    'Order Value': order_values,
    'Discount Percentage': discounts,
    'Delivery Time (Days)': delivery_times,
    'Customer Rating': customer_ratings,
    'Returned': returned
})

# Display the first 10 rows
display(df.head(10))
```

## Step 3 — Perform Basic Descriptive Analysis

Now ask Gemini to calculate descriptive statistics for the numerical variables.

Try this prompt:

> **Calculate the mean, median, standard deviation, minimum, and maximum for the numerical variables in this dataset. Present the results clearly in a table.**

Paste the code generated by Gemini into the cell below and run it.


```python
# Calculate descriptive statistics for numerical variables
numerical_stats = df.describe().loc[['mean', '50%', 'std', 'min', 'max']]

# Rename '50%' index to 'median' for clarity
numerical_stats = numerical_stats.rename(index={'50%': 'median'})

# Display the formatted table
print("Descriptive Statistics for Numerical Variables:")
display(numerical_stats.round(2))
```

## Step 4 — Ask a Business Question

Descriptive statistics are useful, but a business analyst should also ask questions that can support decision-making.

Ask Gemini:

> **Calculate the average order value for each product category. Identify the category with the highest average order value and present the results clearly.**

Paste and run the generated code below.


```python
# Group by product category and calculate mean order value
category_avg_value = df.groupby('Product Category')['Order Value'].mean().sort_values(ascending=False).reset_index()

# Rename columns for clarity
category_avg_value.columns = ['Product Category', 'Average Order Value ($)']

# Identify the category with the highest average order value
highest_category = category_avg_value.iloc[0]

# Present the results clearly
print("Average Order Value by Product Category:")
display(category_avg_value.round(2))
print(f"\nThe category with the highest average order value is '{highest_category['Product Category']}' with an average order value of ${highest_category['Average Order Value ($)']:,.2f}.")
```

## Step 5 — Create a Visualization

Now ask Gemini to visualize one of the results.

Try this prompt:

> **Create an appropriate visualization showing the average order value by product category. Include a clear title and labels for the axes.**

Paste and run the generated code below.

Take a moment to look at the visualization and check whether it communicates the result clearly.


```python
import matplotlib.pyplot as plt
import seaborn as sns

# Set the visual style
sns.set_theme(style="whitegrid")
plt.figure(figsize=(10, 6))

# Create a bar plot (updated to avoid deprecation warning)
ax = sns.barplot(
    x='Product Category',
    y='Average Order Value ($)',
    data=category_avg_value,
    hue='Product Category',
    palette='Blues_r',
    legend=False
)

# Add values on top of the bars
for container in ax.containers:
    ax.bar_label(container, fmt='$%.2f', padding=3)

# Add title and labels
plt.title('Average Order Value by Product Category', fontsize=16, pad=15, fontweight='bold')
plt.xlabel('Product Category', fontsize=12)
plt.ylabel('Average Order Value ($)', fontsize=12)

# Display the plot
plt.tight_layout()
plt.show()
```

```python
# Calculate the return rate by product category
# Since 'Returned' is a boolean (True/False), the mean represents the proportion of True values.
return_rates = df.groupby('Product Category')['Returned'].mean().sort_values(ascending=False).reset_index()

# Convert to percentage and rename columns for clarity
return_rates['Return Rate (%)'] = (return_rates['Returned'] * 100).round(2)
return_rates = return_rates.drop(columns=['Returned'])

# Display the sorted return rates
print("Return Rate by Product Category:")
display(return_rates)

# Identify the category with the highest return rate
highest_return = return_rates.iloc[0]
print(f"\nThe category with the highest return rate is '{highest_return['Product Category']}' with {highest_return['Return Rate (%)']}% of orders returned.")
```
